```python
# Instala a biblioteca necessária (rode apenas uma vez por sessão)
!pip install mlxtend -q

import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# Carrega o dataset de uma posição específica
# Altere o nome do arquivo conforme a posição que está analisando agora
df_posicao = pd.read_excel('Z_DISCRETIZAÇÃO_GERAL.xlsx')

print(f"Dataset carregado com {len(df_posicao)} instâncias.")
display(df_posicao.head(3))
```

```python
# Transforma todas as colunas textuais em colunas booleanas (Verdadeiro/Falso)
df_encoded = pd.get_dummies(df_posicao).astype(bool)

print(f"Colunas originais: {len(df_posicao.columns)}")
print(f"Colunas após a transformação (One-Hot): {len(df_encoded.columns)}")
```

```python
from mlxtend.frequent_patterns import apriori, association_rules
from google.colab import files

# ============================================================
# 1. Define os parâmetros do teste
# ============================================================

suporte_teste = 0.010      # 10% de suporte
confianca_teste = 0.50    # 60% de confiança
lift_minimo = 1.20        # Filtra apenas regras com lift maior que 1.20

# ============================================================
# 2. Aplica o Apriori
# ============================================================

itens_frequentes = apriori(
    df_encoded,
    min_support=suporte_teste,
    use_colnames=True,
    max_len=3
)

# Verifica se encontrou itens frequentes
if itens_frequentes.empty:
    print("Nenhum item frequente encontrado com esse suporte.")

else:
    # ============================================================
    # 3. Gera as regras de associação
    # ============================================================

    regras_gerais = association_rules(
        itens_frequentes,
        metric="confidence",
        min_threshold=confianca_teste
    )

    # Verifica se encontrou regras
    if regras_gerais.empty:
        print("Nenhuma regra encontrada com essa confiança.")

    else:
        # ============================================================
        # 4. Função para garantir que o antecedente NÃO tenha IS
        # ============================================================

        def antecedente_sem_is(antecedents):
            return not any(str(item).startswith('IS_') for item in antecedents)

        # ============================================================
        # 5. Função para garantir que o consequente seja APENAS um IS
        # ============================================================

        def consequente_apenas_is(consequents):
            return len(consequents) == 1 and list(consequents)[0] in [
                'IS_POSITIVO',
                'IS_NEGATIVO',
                'IS_NEUTRO'
            ]

        # ============================================================
        # 6. Filtra regras no formato:
        # estatísticas de desempenho -> índice de sentimento
        # ============================================================

        regras_focadas_no_is = regras_gerais[
            regras_gerais['antecedents'].apply(antecedente_sem_is) &
            regras_gerais['consequents'].apply(consequente_apenas_is)
        ]

        # ============================================================
        # 7. Filtra apenas regras com lift > 1.20
        # ============================================================

        regras_focadas_no_is = regras_focadas_no_is[
            regras_focadas_no_is['lift'] > lift_minimo
        ]

        # ============================================================
        # 8. Ordena por lift, confiança e suporte
        # ============================================================

        regras_focadas_no_is = regras_focadas_no_is.sort_values(
            by=['lift', 'confidence', 'support'],
            ascending=[False, False, False]
        )

        # ============================================================
        # 9. Seleciona apenas as colunas principais
        # ============================================================

        resultado = regras_focadas_no_is[
            ['antecedents', 'consequents', 'support', 'confidence', 'lift']
        ]

        print(f"\nRegras no formato estatísticas -> IS com lift > {lift_minimo}: {len(resultado)}")

        # Mostra apenas as 20 primeiras no Colab
        display(resultado.head(20))

        # ============================================================
        # 10. Salva TODAS as regras filtradas em Excel
        # ============================================================

        nome_arquivo = "GERAL_regras_apriori_suporte_0010_confianca_050.xlsx"

        resultado.to_excel(nome_arquivo, index=False)

        print(f"\nArquivo salvo com sucesso: {nome_arquivo}")

        # Faz download do arquivo
        files.download(nome_arquivo)
```
